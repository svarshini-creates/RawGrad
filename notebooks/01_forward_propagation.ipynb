# RawGrad — Forward Propagation

This notebook demonstrates forward propagation using the RawGrad neural-network framework.

We will build a small neural network from scratch and trace how input data moves through:

- Dense layers
- ReLU activation
- Softmax activation
- A complete `Sequential` model

The goal is to understand what a neural network does during the forward pass before introducing backpropagation.

```python
import sys
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT / "src"))

from rawgrad.layers import Dense
from rawgrad.activations import ReLU, Softmax
from rawgrad.model import Sequential
```

```python
np.random.seed(42)

X = np.array([
    [1.0, 2.0, 3.0],
    [2.0, 1.0, 0.5],
    [0.0, 1.0, 2.0],
    [3.0, 2.0, 1.0],
])

print("Input data:")
print(X)
print("Input shape:", X.shape)
```

```python
dense = Dense(
    input_size=3,
    output_size=4,
    initialization="xavier",
)

dense_output = dense.forward(X)

print("Dense output shape:", dense_output.shape)
print("Dense output:")
print(dense_output)
```

```python
relu = ReLU()

relu_output = relu.forward(dense_output)

print("ReLU output shape:", relu_output.shape)
print("ReLU output:")
print(relu_output)
```

```python
model = Sequential([
    Dense(3, 8, initialization="xavier"),
    ReLU(),
    Dense(8, 3, initialization="xavier"),
    Softmax(),
])

predictions = model.forward(X)

print("Model output shape:", predictions.shape)
print("Model predictions:")
print(predictions)
```

```python
probability_sums = np.sum(predictions, axis=1)
predicted_classes = np.argmax(predictions, axis=1)

print("Probability sums:", probability_sums)
print("Predicted classes:", predicted_classes)
```

```python
assert predictions.shape == (4, 3)
assert np.allclose(probability_sums, 1.0)
assert np.all(np.isfinite(predictions))

print("FORWARD PROPAGATION: PASS")
```
